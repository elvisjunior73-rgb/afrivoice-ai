## 1. Collecte Audio (YouTube)

```python
import os
import yt_dlp

def download_youtube_audio(url, language, output_dir):
    os.makedirs(output_dir, exist_ok=True)
    ydl_opts = {
        'format': 'bestaudio/best',
        'postprocessors': [{
            'key': 'FFmpegExtractAudio',
            'preferredcodec': 'wav',
            'preferredquality': '192',
        }],
        'postprocessor_args': ['-ar', '16000', '-ac', '1'],
        'outtmpl': os.path.join(output_dir, f'%(id)s_{language}.%(ext)s'),
    }
    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        ydl.download([url])

# Exemple d'utilisation
# download_youtube_audio('URL_YOUTUBE', 'lin', 'donnees/brutes/lin')
```

## 2. Segmentation et Transcription (Whisper)

```python
import whisper
import json

def transcribe_audio(audio_file, output_dir, model_size="base"):
    os.makedirs(output_dir, exist_ok=True)
    model = whisper.load_model(model_size)
    result = model.transcribe(audio_file)
    
    output_file = os.path.join(output_dir, "transcription.json")
    with open(output_file, 'w', encoding='utf-8') as f:
        json.dump(result, f, ensure_ascii=False, indent=4)
    print(f"Transcription sauvegardée dans {output_file}")

# Exemple d'utilisation
# transcribe_audio('donnees/brutes/lin/audio.wav', 'donnees/transcriptions/lin')
```
